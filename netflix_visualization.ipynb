```python
#Loading the netflix dataset from Kaggle
import pandas as pd
dataset=pd.read_csv("netflix_dataset.csv")
print(dataset.head())
```

```python
#Creating Dataframe
df=pd.DataFrame(dataset)
df.head()
```

```python
# knowing about the shape of the dataset(rows and columns)
df.shape
```

```python
df.info()
```

```python
df.describe(include='object')
```

```python
df['director']=df['director'].fillna('unknown')
```

```python
df.describe(include=object)
```

```python
#handling missing values by fillna()
```

```python
df['cast']=df['cast'].fillna('unknown')
```

```python
df['country']=df['country'].fillna('not mentioned')
```

```python
df['rating']=df['rating'].fillna('Not rated')
```

```python
df['duration']=df['duration'].fillna('not mentioned')
```

```python
df['listed_in']=df['listed_in'].fillna('not mentioned')
```

```python
df['description']=df['description'].fillna('not available')
```

```python
df.isnull().sum()
```

```python
df=df.dropna(subset=['date_added'])
```

```python
df.isnull().sum()
```

```python
df['date_added']=pd.to_datetime(df['date_added'],errors='coerce')
```

```python
df.describe(include='object')
```

```python
df.shape
```

```python
df['date_added'].head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.countplot(data=df,x='type',order=['TV Show','Movie'],orient='v',palette='Set3')
plt.title("Movies vs TV shows")
plt.xlabel("Type")
plt.ylabel("Count")
plt.show()
```

```python
sns.countplot(data=df,y='rating',order=df['rating'].value_counts(ascending=True).index,palette='coolwarm')
plt.title("Content Ratings")
plt.xlabel("Count")
plt.ylabel('Rating')
plt.show()
```

```python
df['rating'].value_counts().reset_index()
```

```python
df['rating'].value_counts(ascending=True)
```

```python
from collections import Counter
genres=df['listed_in'].dropna().apply(lambda x: x.split(','))
flat_genres=[g.strip() for sublist in genres for g in sublist]
top_genres=Counter(flat_genres).most_common(10)
genres,counts=zip(*top_genres)
sns.barplot(x=list(counts),y=list(genres),palette='viridis')
plt.title("Top 10 Most Common Genres on Netflix")
plt.xlabel("Number of Titles")
plt.ylabel("Genres")
plt.show()
```

```python
df['date_added']=pd.to_datetime(df['date_added'],errors='coerce')
df['date_added'].unique()
```

```python
df=df.dropna(subset=['date_added'])
```

```python
df['year_added']=df['date_added'].dt.year
df['year_added'].unique()
```

```python
df['month_added']=df['date_added'].dt.month_name
df['month_added'].unique()
```

```python
df['month_added']=df['date_added'].dt.month_name()
monthly=df['month_added'].value_counts().sort_index()
monthly
```

```python
yearly=df['year_added'].value_counts().sort_index()
sns.lineplot(x=yearly.index,y=yearly.values,marker='o')
plt.title("Number of titles Added per year")
plt.xlabel("Year")
plt.ylabel("Counts")
plt.grid(True)
plt.show()
```

```python
yearly
```

```python
fig=plt.figure(figsize=(10,8))
sns.lineplot(x=monthly.index,y=monthly.values,marker='x')
plt.title("NUmber of titles added by month")
plt.xlabel("Month")
plt.ylabel("Count")
plt.show()
```

```python
# here we use countplot instead of lineplot because we can see clearly the points
sns.countplot(data=df,y='month_added',order=df['month_added'].value_counts().index,palette='cubehelix')
plt.title("Number of titles added by month")
plt.xlabel('Counts')
plt.ylabel('Month')
plt.show()
```
